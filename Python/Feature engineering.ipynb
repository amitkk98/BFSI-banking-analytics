```python
transaction_sorted= transaction.sort_values('Transaction_Date')
```

```python
transaction_sorted.head()
```

```python
#latest balance for each customer
latest_account_balance= (transaction_sorted.groupby(['Customer_ID','Account_Type']).last())
```

```python
latest_account_balance.head()
```

```python
latest_account_balance =latest_account_balance.groupby('Customer_ID')['Total_Balance'].sum()
```

```python
latest_account_balance.head()
```

```python
#latest investment amount per customer
latest_investment=transaction_sorted.groupby('Customer_ID')['Investment_Amount'].last()
```

```python
latest_investment.head()
```

```python
net_worth = pd.DataFrame({
    'Total_Account_Balance': latest_account_balance,
    'Current_Investment': latest_investment
})
```

```python
net_worth.head()
```

```python
net_worth['Net_Worth']= net_worth['Total_Account_Balance']+net_worth['Current_Investment']
```

```python
net_worth.head()
```

```python
customer=customer.merge(
    net_worth,
    on='Customer_ID',
    how='left'
)
```

```python
customer.head()
```

```python
customer.describe()
```

```python
customer=customer.fillna(0)
```

```python
customer.head()
```

```python
customer['Customer_Type'] =pd.cut(
    customer['Net_Worth'],
    bins=[float('-inf'),0,75000,200000,float('inf')],
    labels=['Inactive', 'Regular', 'Premium', 'Gold']
)
```

```python
#result after cou=stomer type column
customer.head()
```

```python
#calculating transaction count per customer
transaction.head()
```

```python
#calculating Transaction_Count per customer
#calculating total transaction anount
# calculating last transacttion date.

customer_activity =(
    transaction.groupby('Customer_ID').agg(
        Transaction_Count=('Transaction_ID','nunique'),
        Total_Transaction_Amount= ('Transaction_Amount','sum'),
        Last_Transaction_Date=('Transaction_Date','max')
    )
)
```

```python
customer_activity.head()
```

```python
customer_feature=customer.merge(
    customer_activity,
    on= 'Customer_ID',
    how= 'left'
)
```

```python
customer_feature.head()
```

```python
customer_feature['Transaction_Count']=(customer_feature['Transaction_Count'].fillna(0).astype(int))
```

```python
customer_feature['Total_Transaction_Amount'] = (customer_feature['Total_Transaction_Amount'].fillna(0))
```

```python
customer_feature.head()
```
